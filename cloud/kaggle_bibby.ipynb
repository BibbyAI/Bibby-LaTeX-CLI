## Write a document

```python
import bibby_cloud

bibby_cloud.write_tex("main.tex", r"""
\documentclass[11pt]{article}
\usepackage[margin=1in]{geometry}
\usepackage{amsmath, amssymb, booktabs, graphicx}
\usepackage[hidelinks]{hyperref}

\title{Written in a notebook, typeset by LaTeX}
\author{You}
\begin{document}
\maketitle

\section{It compiles}
\begin{equation}
  \nabla_\theta J(\theta) = \mathbb{E}\left[ \nabla_\theta \log \pi_\theta(a|s)\, A(s,a) \right]
\end{equation}

\begin{table}[h]\centering
\begin{tabular}{lr}\toprule Run & Score \\ \midrule A & 0.91 \\ B & 0.88 \\ \bottomrule\end{tabular}
\caption{A table.}
\end{table}

\end{document}
""")
```

## Compile and preview

```python
# compile and preview the PDF right here
bibby_cloud.compile_and_show("main.tex")
```
